# Listas


- Colección **ordenada** y **mutable** de elementos
- Los índices empiezan en **0**
- Soportan **slicing**: `lista[inicio:fin:paso]`

````python
frutas = ["manzana", "pera", "uva", "kiwi", "mango"]

print(frutas[0])    # primer elemento
print(frutas[-1])   # último elemento
print(frutas[1:3])  # slice: pera, uva
print(frutas[::2])  # cada 2 elementos

# Métodos útiles
frutas.append("naranja")   # agrega al final
frutas.insert(1, "ciruela") # inserta en posición
frutas.remove("pera")       # elimina por valor
frutas.sort()               # ordena in-place

print(f"\nLista final: {frutas}")
print(f"Largo: {len(frutas)}")
````

# Diccionarios


- Colección de pares **llave → valor**
- Acceso por llave, no por índice numérico
- Las llaves deben ser **únicas** e **inmutables**

````python
alumno = {
    "nombre": "Carlos",
    "edad":    22,
    "carrera": "Ingeniería",
    "notas":   [6.5, 5.8, 7.0]
}

# Acceder a valores
print(alumno["nombre"])
print(alumno["notas"][0])   # primera nota

# Agregar y modificar
alumno["email"] = "carlos@mail.com"
alumno["edad"]  = 23

# Iterar
for llave, valor in alumno.items():
    print(f"  {llave}: {valor}")
````

# Salida de Datos — `print()`


Hay varias formas de formatear la salida:

````python
nombre = "María"
edad   = 30
altura = 1.65

# 1. Concatenación (poco recomendado)
print("Nombre: " + nombre + ", Edad: " + str(edad))

# 2. Comas (sep por defecto es espacio)
print("Nombre:", nombre, "Edad:", edad, sep=" | ")

# 3. % formatting (estilo antiguo)
print("Altura: %.2f m" % altura)

# 4. f-strings (RECOMENDADO — Python 3.6+)
print(f"Hola {nombre}, tienes {edad} años y mides {altura:.2f} m")

# Parámetros útiles de print
print("línea 1", end=" → ")
print("línea 2")
````

# Operadores de Comparación y Lógicos


**Comparación** — retornan `bool`:

| Operador | Significado |
|----------|-------------|
| `==` | Igual a |
| `!=` | Distinto de |
| `>` / `<` | Mayor / Menor |
| `>=` / `<=` | Mayor o igual / Menor o igual |

**Lógicos:**

| Operador | Descripción | Ejemplo |
|----------|-------------|---------|
| `and` | Ambos verdaderos | `x > 0 and x < 10` |
| `or` | Al menos uno verdadero | `x == 0 or x == 1` |
| `not` | Negación | `not True → False` |

````python
x = 7

print(x > 5)            # True
print(x == 7)           # True
print(x > 5 and x < 10) # True  (ambas se cumplen)
print(x < 0 or x > 5)  # True  (al menos una)
print(not (x == 7))     # False (negación)
````

# Condicional `if / elif / else`


```python
if <condición1>:
    # se ejecuta si condición1 es True
elif <condición2>:
    # se ejecuta si condición2 es True (y condición1 fue False)
else:
    # se ejecuta si ninguna condición fue True
```
> La **indentación** (4 espacios) es **obligatoria** en Python — define los bloques

````python
nota = 5.8

if nota >= 6.0:
    print("Aprobado ✅")
elif nota >= 4.0:
    print("En recuperación 🔶")
else:
    print("Reprobado ❌")

print(f"Nota: {nota}")
````

# Estructura de Control: `for`


- Itera sobre cualquier secuencia: listas, strings, rangos, diccionarios...
- `range(inicio, fin, paso)` genera secuencias de enteros

````python
# Iterar sobre lista
colores = ["rojo", "verde", "azul"]
for color in colores:
    print(f"Color: {color}")

print()

# Iterar con índice usando enumerate
for i, color in enumerate(colores):
    print(f"[{i}] {color}")

print()

# range: contar del 1 al 5
for n in range(1, 6):
    print(f"{n}² = {n**2}")
````

# Estructura de Control: `while`


- Ejecuta un bloque **mientras** la condición sea `True`
- Siempre verificar que la condición eventualmente se vuelva `False` para evitar bucles infinitos
- `break` sale del loop; `continue` salta a la siguiente iteración

````python
# Ejemplo: validar entrada del usuario
# (En práctica, ejecutar y escribir distintos valores)
intentos = 0
contrasena_correcta = "ds2024"

while intentos < 3:
    clave = input("Ingresa la contraseña: ")
    if clave == contrasena_correcta:
        print("Acceso concedido ✅")
        break
    else:
        intentos += 1
        print(f"Contraseña incorrecta. Intentos restantes: {3 - intentos}")
else:
    print("Cuenta bloqueada ❌")
````

````python
# Ejemplo: múltiplos de 3 hasta 30
print("Múltiplos de 3 hasta 30:")
n = 1
while n <= 30:
    if n % 3 == 0:
        print(n, end=" ")
    n += 1
````

# List Comprehensions *(bonus)*


Forma compacta y pythónica de construir listas:
```python
nueva_lista = [expresión for elemento in iterable if condición]
```

````python
# Forma tradicional
cuadrados = []
for n in range(1, 11):
    cuadrados.append(n ** 2)

# Con list comprehension (equivalente)
cuadrados = [n**2 for n in range(1, 11)]
print(cuadrados)

# Con filtro: solo pares
pares = [n for n in range(1, 21) if n % 2 == 0]
print(pares)

# Transformar strings
nombres = ["ana", "luis", "maría"]
mayus   = [n.capitalize() for n in nombres]
print(mayus)
````

### Problema: Analizador de notas

Escribe un programa que:

1. Pida al usuario **5 notas** (de 1.0 a 7.0) usando `input()` dentro de un `for`
2. Las almacene en una lista
3. Calcule e imprima:
   - El **promedio**
   - La nota **más alta** y la **más baja** (sin usar `max()`/`min()` directamente — usa un `for`)
   - Cuántas notas son **aprobadas** (≥ 4.0)
4. Entregue un veredicto final:
   - Promedio ≥ 5.5 → `"Excelente"`
   - Promedio ≥ 4.0 → `"Aprobado"`
   - Promedio < 4.0 → `"Reprobado"`

````python
# Actividad Final — Analizador de notas

notas = []

# 1. Ingresar 5 notas
for i in range(1, 6):
    while True:
        try:
            nota = float(input(f"Ingresa nota {i} (1.0 - 7.0): "))
            if 1.0 <= nota <= 7.0:
                notas.append(nota)
                break
            else:
                print("La nota debe estar entre 1.0 y 7.0")
        except ValueError:
            print("Ingresa un número válido")

# 2. Calcular estadísticas
promedio  = sum(notas) / len(notas)
mayor     = notas[0]
menor     = notas[0]
aprobadas = 0

for nota in notas:
    if nota > mayor:
        mayor = nota
    if nota < menor:
        menor = nota
    if nota >= 4.0:
        aprobadas += 1

# 3. Veredicto
if promedio >= 5.5:
    veredicto = "Excelente 🌟"
elif promedio >= 4.0:
    veredicto = "Aprobado ✅"
else:
    veredicto = "Reprobado ❌"

# 4. Mostrar resultados
print(f"\n{'='*35}")
print(f"  Notas:     {notas}")
print(f"  Promedio:  {promedio:.2f}")
print(f"  Más alta:  {mayor}")
print(f"  Más baja:  {menor}")
print(f"  Aprobadas: {aprobadas}/5")
print(f"  Veredicto: {veredicto}")
print(f"{'='*35}")
````
