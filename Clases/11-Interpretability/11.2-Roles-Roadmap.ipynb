# 1. Quiz: descubri tu perfil

Respondé las preguntas honestamente. Al final el codigo te dice los TOP 3 roles que mas se ajustan a vos.

````python
# QUIZ — cambia los valores segun tu honestidad
# Cada respuesta: 1 (nada), 2 (poco), 3 (medio), 4 (mucho), 5 (apasionado)

mis_respuestas = {
    'me_apasiona_modelar': 4,             # entrenar modelos, comparar metricas
    'disfruto_codear_apis': 3,            # FastAPI, endpoints, docker
    'me_atraen_pipelines': 3,             # ETL, SQL, transformaciones de datos
    'me_volaron_los_llms': 5,             # GPT, Claude, MATEO, RAG, agentes
    'me_gustan_los_dashboards': 2,        # Power BI, Tableau, Looker
    'amo_el_devops': 3,                   # CI/CD, Kubernetes, monitoring
    'me_interesa_la_estrategia': 4,       # producto, priorizacion, stakeholders
    'tolero_codigo_legacy': 2,            # arreglar codigo viejo, debugging
    'aprendo_rapido_herramientas_nuevas': 5,
    'me_motiva_explicar_a_no_tecnicos': 4,
}

# Mapeo: que respuesta pesa para que rol (10 dimensiones por rol)
roles_pesos = {
    'DS tradicional':    [4, 1, 1, 1, 3, 1, 2, 2, 2, 4],
    'ML Engineer':       [3, 5, 2, 2, 1, 4, 2, 3, 4, 2],
    'Data Engineer':     [1, 2, 5, 1, 1, 4, 1, 3, 3, 2],
    'GenAI Engineer':    [3, 4, 2, 5, 1, 3, 2, 2, 5, 3],
    'Analytics Eng':     [2, 2, 4, 1, 4, 2, 3, 3, 3, 4],
    'MLOps Engineer':    [2, 4, 3, 1, 1, 5, 1, 4, 3, 1],
    'AI Product Manager':[3, 2, 1, 4, 4, 1, 5, 2, 4, 5],
}

import pandas as pd
preguntas = list(mis_respuestas.keys())
scores = {rol: sum(mis_respuestas[p] * w for p, w in zip(preguntas, pesos)) for rol, pesos in roles_pesos.items()}

ranking = pd.DataFrame(sorted(scores.items(), key=lambda x: -x[1]), columns=['Rol', 'Match Score'])
ranking.index = range(1, len(ranking)+1)
print('=== TU RANKING ===')
print(ranking.to_string())
print()
print(f'Rol con mas afinidad: {ranking.iloc[0]["Rol"]}')
print(f'Tus 3 caminos posibles: {", ".join(ranking.iloc[:3]["Rol"].tolist())}')
````

**Como usar el quiz**: cambia los numeros en `mis_respuestas` por TU evaluacion honesta (1 a 5) y re-corre la celda. El ranking te dice cuales 3 roles deberias considerar.

# 2. Visualizacion: salarios y demanda por rol en LATAM (2026)

````python
import matplotlib.pyplot as plt
import numpy as np

roles = ['DS tradicional', 'ML Engineer', 'Data Engineer', 'GenAI Engineer',
         'Analytics Eng', 'MLOps Engineer', 'AI Product Manager']
salario_min = [1500, 2500, 2500, 3000, 2000, 3000, 3500]
salario_max = [3500, 5000, 4500, 6500, 4000, 5500, 6000]
demanda = ['Alta', 'Muy Alta', 'Muy Alta', 'Muy Alta', 'Alta', 'Alta', 'Creciente']

color_map = {'Alta': '#4472C4', 'Muy Alta': '#C0504D', 'Creciente': '#FFC000'}
colors = [color_map[d] for d in demanda]

fig, ax = plt.subplots(figsize=(13, 6))
y_pos = np.arange(len(roles))
width = [smax - smin for smin, smax in zip(salario_min, salario_max)]
ax.barh(y_pos, width, left=salario_min, color=colors, edgecolor='white', alpha=0.85)

for i, (smin, smax, d) in enumerate(zip(salario_min, salario_max, demanda)):
    ax.text(smin - 100, i, f'${smin:,}', va='center', ha='right', fontsize=9, color='gray')
    ax.text(smax + 100, i, f'${smax:,}', va='center', ha='left', fontsize=9, fontweight='bold')
    ax.text((smin + smax)/2, i, d, va='center', ha='center', fontsize=10, color='white', fontweight='bold')

ax.set_yticks(y_pos); ax.set_yticklabels(roles, fontsize=11)
ax.invert_yaxis()
ax.set_xlabel('Salario mensual mid-level USD (LATAM)')
ax.set_title('Rango salarial y demanda por rol DS - 2026', fontweight='bold', fontsize=14)
ax.set_xlim(0, 7500)

from matplotlib.patches import Patch
legend_elements = [Patch(facecolor=c, label=d, alpha=0.85) for d, c in color_map.items()]
ax.legend(handles=legend_elements, loc='lower right', title='Demanda')

plt.tight_layout(); plt.show()
````

```python
from pydantic import BaseModel
import mlflow

# 1. Cargar modelo desde el registry
model = mlflow.sklearn.load_model("models:/churn-prod/Production")

# 2. Definir el contrato de la API
class Features(BaseModel):
    tenure: int
    monthly_charges: float
    contract_type: str

# 3. Exponer endpoint
app = FastAPI()

@app.post("/predict")
async def predict(features: Features):
    prediction = model.predict_proba([[features.tenure, features.monthly_charges]])
```

## GenAI Engineer — un dia tipico

```python
from langchain_anthropic import ChatAnthropic
from langchain.vectorstores import Chroma
from langchain.chains import RetrievalQA

vectorstore = Chroma(persist_directory="./kb", embedding_function=OpenAIEmbeddings())
llm = ChatAnthropic(model="claude-sonnet-4-6", temperature=0.1)

qa = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=vectorstore.as_retriever(search_kwargs={"k": 5}),

async def chat(message: str):
```

## Analytics Engineer — un dia tipico

Modelar capa semantica con dbt:

```sql
-- models/marts/finance/customer_metrics.sql

SELECT
    c.customer_id,
    DATEDIFF('month', c.signup_date, CURRENT_DATE) AS tenure_months,
    SUM(s.amount) AS total_lifetime_value,
    p.churn_probability,
    CASE
        WHEN p.churn_probability > 0.7 THEN 'High Risk'
        WHEN p.churn_probability > 0.4 THEN 'Medium Risk'
        ELSE 'Low Risk'
    END AS risk_segment
FROM {{ ref('stg_customers') }} c
LEFT JOIN {{ ref('fct_subscriptions') }} s USING (customer_id)
LEFT JOIN {{ ref('ml_predictions_daily') }} p USING (customer_id)
GROUP BY c.customer_id, p.churn_probability
```

Stack: dbt, SQL, BigQuery/Snowflake, Looker/Tableau/Power BI, Git.
