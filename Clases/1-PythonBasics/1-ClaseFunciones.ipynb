# ¿Qué es una función?


Una función es un **bloque de código reutilizable** con nombre. Piénsala como una **receta**:
- La *receta* tiene un nombre, ingredientes (parámetros) y produce un resultado (return)
- Puedes usarla cuantas veces quieras sin reescribir el código

```python
def nombre_funcion(parametro1, parametro2):
    # código
    return resultado
```

> **¿Por qué usar funciones?**
> - ♻️ No repetir código (principio DRY: Don't Repeat Yourself)
> - 🧩 Dividir problemas grandes en partes pequeñas
> - 🧪 Facilitar pruebas y corrección de errores

# Función básica con `return`


````python
# Función que calcula el área de un rectángulo
def area_rectangulo(base, altura):
    area = base * altura
    return area

# Llamar (invocar) la función
resultado = area_rectangulo(5, 3)
print(f"Área: {resultado}")

# También se puede usar directamente
print(f"Área: {area_rectangulo(10, 4)}")
print(f"Área: {area_rectangulo(2.5, 8)}")
````

# Función sin `return` (procedimiento)


````python
# No toda función necesita retornar un valor
# Algunas solo ejecutan acciones

def saludar(nombre, hora):
    if hora < 12:
        saludo = "Buenos días"
    elif hora < 20:
        saludo = "Buenas tardes"
    else:
        saludo = "Buenas noches"
    print(f"{saludo}, {nombre}! 👋")

saludar("Ana", 10)
saludar("Luis", 15)
saludar("María", 21)

# Si no hay return, Python retorna None implícitamente
resultado = saludar("Pedro", 9)
print(f"Retorna: {resultado}")   # None
````

# Parámetros por defecto


````python
# Los parámetros con default son opcionales al llamar la función
def calcular_descuento(precio, descuento=0.10, incluir_iva=True):
    precio_con_descuento = precio * (1 - descuento)
    if incluir_iva:
        precio_final = precio_con_descuento * 1.19
    else:
        precio_final = precio_con_descuento
    return round(precio_final, 2)

# Llamadas con distintas combinaciones
print(calcular_descuento(10_000))               # usa defaults: 10% desc + IVA
print(calcular_descuento(10_000, 0.20))         # 20% desc + IVA
print(calcular_descuento(10_000, 0.15, False))  # 15% desc sin IVA
print(calcular_descuento(10_000, incluir_iva=False))  # solo cambia IVA
````

# Múltiples valores de retorno


````python
# Python permite retornar varios valores como tupla
def estadisticas(numeros):
    total    = sum(numeros)
    promedio = total / len(numeros)
    maximo   = max(numeros)
    minimo   = min(numeros)
    return promedio, maximo, minimo   # retorna tupla

notas = [6.5, 4.8, 7.0, 5.5, 6.1]

# Desempaquetar el resultado
prom, alto, bajo = estadisticas(notas)
print(f"Promedio: {prom:.2f}")
print(f"Más alta: {alto}")
print(f"Más baja: {bajo}")

# O guardar la tupla completa
resultado = estadisticas(notas)
print(f"\nTodo junto: {resultado}")
````

# Funciones que llaman a otras funciones


````python
# Las funciones pueden llamarse entre sí — así se construyen programas grandes

def es_par(n):
    return n % 2 == 0

def clasificar_numero(n):
    if n == 0:
        return "cero"
    elif n < 0:
        return "negativo"
    elif es_par(n):         # llama a otra función
        return "par positivo"
    else:
        return "impar positivo"

def analizar_lista(numeros):
    resultado = {}
    for n in numeros:
        resultado[n] = clasificar_numero(n)  # llama a otra función
    return resultado

lista = [-3, 0, 4, 7, -8, 15, 2]
for numero, categoria in analizar_lista(lista).items():
    print(f"  {numero:4d} → {categoria}")
````

# Scope: variables locales vs globales


````python
# Las variables dentro de una función son LOCALES — no existen fuera

contador_global = 0  # variable global

def incrementar():
    local_var = 100  # solo existe dentro de la función
    return local_var

print(incrementar())   # 100
# print(local_var)     # ❌ NameError: local_var no existe aquí

# Para modificar una global dentro de una función (poco recomendado)
def resetear():
    global contador_global
    contador_global = 0

contador_global = 5
print(f"Antes:  {contador_global}")
resetear()
print(f"Después: {contador_global}")

# ✅ Mejor práctica: pasar y retornar valores
def incrementar_v2(contador, cantidad=1):
    return contador + cantidad

c = 0
c = incrementar_v2(c)
c = incrementar_v2(c, 5)
print(f"Contador: {c}")
````

# ¿Qué es una librería?


Una **librería** es un conjunto de funciones y clases escritas por alguien más que podemos reutilizar.

```python
import libreria                    # importar toda la librería
import libreria as alias           # con alias (más corto)
from libreria import funcion       # importar solo una función
from libreria import func1, func2  # importar varias
```

> 📦 Las librerías que vienen con Python se llaman **stdlib** (standard library).  
> Las externas como `pandas` se instalan con `pip install nombre`.

# `math` — Matemáticas


````python
import math

print(f"π         = {math.pi:.6f}")
print(f"e         = {math.e:.6f}")
print(f"√144      = {math.sqrt(144)}")
print(f"log₁₀(100)= {math.log10(100)}")
print(f"log₂(8)   = {math.log2(8)}")
print(f"ceil(4.1) = {math.ceil(4.1)}")   # redondear hacia arriba
print(f"floor(4.9)= {math.floor(4.9)}")  # redondear hacia abajo
print(f"abs(-7)   = {abs(-7)}")          # valor absoluto (builtin)
print(f"2^10      = {math.pow(2, 10):.0f}")
print(f"sin(90°)  = {math.sin(math.radians(90)):.2f}")
````

# `random` — Aleatoriedad


````python
import random

random.seed(42)  # fija la semilla — resultados reproducibles

print(f"Float [0,1):     {random.random():.4f}")
print(f"Entero [1,10]:   {random.randint(1, 10)}")
print(f"Float [5, 15]:   {random.uniform(5, 15):.2f}")

opciones = ["Rock", "Papel", "Tijera"]
print(f"choice:          {random.choice(opciones)}")

lista = list(range(1, 11))
random.shuffle(lista)
print(f"shuffle:         {lista}")

muestra = random.sample(range(1, 50), 6)
print(f"Loto (6/49):     {sorted(muestra)}")
````

# `datetime` — Fechas y horas


````python
from datetime import datetime, date, timedelta

# Fecha y hora actual
ahora = datetime.now()
print(f"Ahora:        {ahora}")
print(f"Fecha:        {ahora.date()}")
print(f"Hora:         {ahora.time()}")
print(f"Año:          {ahora.year}")
print(f"Día semana:   {ahora.strftime('%A')}")  # nombre del día
print(f"Formato:      {ahora.strftime('%d/%m/%Y %H:%M')}")

# Operaciones con fechas
hoy = date.today()
cumple = date(2025, 12, 25)  # Navidad
dias_para_navidad = (cumple - hoy).days
print(f"\nDías para Navidad 2025: {dias_para_navidad}")

hace_30_dias = hoy - timedelta(days=30)
print(f"Hace 30 días: {hace_30_dias}")
````
